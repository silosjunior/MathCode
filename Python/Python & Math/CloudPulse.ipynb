# Módulo de Disponibilidade (Uptime)

Neste módulo, nosso objetivo é converter uma medida de tempo composta, expressa em dias, horas, minutos e segundos, para sua menor unidade fundamental: o segundo.
Na ciência da computação, essa transformação linear é essencial para padronizar medições de desempenho de servidores. A fórmula para o cálculo do tempo total em segundos é dada por:

$$S_{total} = (D \times 86400) + (H \times 3600) + (M \times 60) + S$$

- **D** é a quantidade de dias.
- **H** é a quantidade de horas.
- **M** é a quantidade de minutos.
- **S** é a quantidade de segundos restantes.

Eis abaixo, uma função em Python que realiza a conversão de grandezas apresentada na equação algébrica:

```python
def converte_para_segundos(d, h, m, s):
    """
    Padroniza a quantidade de dias, horas, minutos e segundos no total em segundos.

    Args:
    - d (int): dias a serem convertidos em segundos.
    - h (int): horas a serem convertidas em segundos.
    - m (int): minutos a serem convertidos em segundos.
    - s (int): segundos somados ao tempo total.

    Returns:
    - S (int): o total do tempo em segundos.
    """
    S = d * 86400 + h * 3600 + m * 60 + s
    return S
```

# Módulo da Termodinâmica

Servidores geram muito calor, e muitas ferramentas de infraestrutura em nuvem globais usam a escala Fahrenheit como padrão, enquanto nós usamos Celsius. A fórmula clássica para essa conversão é:

$$F = \frac{9 \times C}{5} + 32$$

Onde:

- **C** representa a temperatura em graus Celsius.
- **F** representa a temperatura em graus Fahrenheit.

Uma função em Python realiza o cálculo em apenas uma linha de código. Veja:

```python
def converte_celsius_para_fahrenheit(t):
    """
    Converte temperatura Celsius em escala Farenheit.

    Args:
    - t (float): temperatura em Celsius a ser convertida em graus Fahrenheit.

    Returns:
    - F (float): temperatura em graus Fahrenheit.
    """
    F = (9 * t) / 5 + 32
    return F
```

# Módulo de Faturamento (Billing)

Na engenharia de nuvem, o modelo de cobrança geralmente envolve uma taxa fixa de manutenção somada a uma taxa variável de consumo. Supondo que o provedor de nuvem cobre um custo fixo de R$ 5,00 por dia que o servidor fica ligado, mais um custo variável de R$ 0,20 por cada Gigabyte (GB) de rede trafegando, temos a seguinte equação matemática formal para calcular o Custo Total $(C_{total})$:

$$C_{total} = (D * 5) + (G * 0.20)$$

Onde:

- **$D$** representa a quantidade de dias em que o servidor foi utilizado.
- **$5$** representa a taxa fixa pelo uso diário do servidor.
- **$G$** é a quantidade de Gigabytes (GB) consumidos durante seu uso.
- **$0.20$** refere-se à taxa cobrada por cada Gigabyte (GB) de rede trafegado.

# Módulo de Transferência (Backup e Migração de Dados)

Na arquitetura de redes e sistemas distribuídos, a previsão do tempo de transferência de grandes volumes de dados baseia-se na relaçõ clássica da cinemática, onde o Tempo é a razão entre o Espaço (Volume de Dados) e a Velocidade (Banda de Rede).
No entanto, em cenários reais, os dados armazenados e a velocidade da rede frequentemente trafegam em ordens de grandeza diferentes (ex: Terabytes em discos, mas Megabytes por segundo na rede). Para evitar constantes fixas no código (os chamados *Magic Numbers*), modelamos uma equação generalizada baseada no sistema binário.
Sabemos que na computação, cada degrau de magnitude representa uma potência de 2 elevada a 10 ($2^{10} = 1024$). Podemos classificar as medidas por um **Índice de Magnitude ($U$)**:

* $U = 0 \rightarrow$ Bytes (B)
* $U = 1 \rightarrow$ Kilobytes (KB)
* $U = 2 \rightarrow$ Megabytes (MB)
* $U = 3 \rightarrow$ Gigabytes (GB)
* $U = 4 \rightarrow$ Terabytes (TB)

Para calcularmos o **Tempo de Transferência ($T$)**, precisamos encontrar a diferença de degraus entre a unidade do Volume ($U_v$) e a unidade da Rede ($U_r$). Essa diferença ($\Delta U = U_v - U_r$) definirá o nosso **Fator de Conversão Exponencial ($F_c$)**.

A equação matemática generalizada é definida por:

$$T = \frac{V \times 1024^{U_v - U_r}}{R}$$

Onde:

- **$T$** é o tempo total estimado para a transferência, em segundos.
- **$V$** é o volume nominal de dados a ser transferido.
- **$U_v$** é o índice de magnitude da unidade do volume armazenado.
- **$U_r$** é o índice de magnitude da unidade de velocidade da rede.
- **$R$** é a taxa de transferência nominal da rede.

**Aplicação no cenário atual:**
Para o nosso diagnóstico padrão, o usuário informará o volume do banco de dados em **Gigabytes ($U_v = 3$)** e a velocidade da rede em **Megabytes por segundo ($U_r = 2$)**.
Substituindo os índices na nossa equação generalizada, a diferença exponencial será $3 - 2 = 1$. Portanto, a equação se adapta perfeitamente ao nosso escopo resolvendo o fator de conversão como $1024^1$:

$$T = \frac{V \times 1024}{R}$$

**Visualização da fórmula matemática em código Python:**

```python
def calcula_tempo_transferencia(
        tamanho_do_arquivo: int,
        velocidade_rede: float,
    ) -> float:
    """
    Calcula o tempo em segundos para a transferência de um determinado volume de dados.

    Args:
    - tamanho_do_arquivo (int): volume de dados a serem transferidos Gigabytes (GB).
    - velocidade_rede (float): velocidade da rede Megabytes por segundo (MB/s).

    Returns:
    - t (float): tempo em segundos para a conclusão da transferência de dados.
    """
    t = (tamanho_do_arquivo * 1024) / velocidade_rede
    return t
```
