Na matemática, podemos calcular a raiz de um número transformando-a em uma operação de exponenciação, desde que o expoente seja fracionário e positivo. A regra geral define que a base elevada a uma fração tem o seu denominador como índice da raiz e o seu numerador como expoente da base.
A fórmula geral é dada por:

$$a^{\frac{x}{y}} = \sqrt[y]{a^x}$$

Aplicando essa regra ao nosso exemplo prático, queremos descobrir a raiz cúbica de 2:

$$2^{\frac{1}{3}} = \sqrt[3]{2}$$
